```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("Unemployment.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Show column names
print(df.columns)

# First 5 rows
print(df.head())

# Dataset info
print(df.info())

# Statistical summary
print(df.describe())

# Missing values
print(df.isnull().sum())

# Remove missing values
df = df.dropna()

# Histogram
sns.histplot(df['Estimated Unemployment Rate (%)'], kde=True)

plt.title("Unemployment Rate Distribution")

plt.show()

# State-wise unemployment
plt.figure(figsize=(12,6))

sns.barplot(
    x='Region',
    y='Estimated Unemployment Rate (%)',
    data=df
)

plt.xticks(rotation=90)

plt.title("State-wise Unemployment Rate")

plt.show()

# Unemployment over time
plt.figure(figsize=(12,6))

sns.lineplot(
    x='Date',
    y='Estimated Unemployment Rate (%)',
    data=df
)

plt.xticks(rotation=90)

plt.title("Unemployment Rate Over Time")

plt.show()

# Insights
print("Insights:")
print("1. Unemployment increased during COVID-19.")
print("2. Different regions showed different unemployment rates.")
print("3. Unemployment trends changed over time.")
```
